```python
# Open a new file named 'notes.txt' in write ("w") mode
with open("notes.txt", "w") as file:
    file.write("Class 1: Python Fundamentals\n")
    file.write("Class 2: Math and Statistics for AI\n")

print("File created and text successfully saved!")
```

```python
# Open the file in read ("r") mode
with open("notes.txt", "r") as file:
    # Read the entire file content as a single block of text
    content = file.read()

print("File Content")
print(content)
```

```python
with open("notes.txt", "r") as file:
    for line in file:
        print(f"Row item: {line.strip()}")  # .strip() removes blank newline spaces
```

```python
# Open the file in append ("a") mode
with open("notes.txt", "a") as file:
    file.write("Class 3: Machine Learning Algorithms\n")

print("New class appended successfully!")
```
