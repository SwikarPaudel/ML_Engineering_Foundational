```python
#List & Tuples

# Create your schedule as a list

weekly_schedule = [
    "Mon: Python Basics",
    "Tue: Data Types & Variables",
    "Wed: Lists & Tuples",
    "Thu: Control Flow",
    "Fri: Functions",
]

# Indexing & slicing
first_day = weekly_schedule[0]
mid_week = weekly_schedule[1:4]  # Tue–Thu
last_two = weekly_schedule[-2:]  # Thu–Fri
first_day, mid_week, last_two

# Update the list
weekly_schedule.append("Sat: Project Time")
weekly_schedule.insert(1, "Mon Eve: Study Group")
weekly_schedule[2] = "Tue: Numbers & Strings"
 
if "Thu: Control Flow" in weekly_schedule:
    weekly_schedule.remove("Thu: Control Flow")
 
removed = weekly_schedule.pop()  # removes last item
weekly_schedule, removed

for i, item in enumerate(weekly_schedule, start=1):
    print(f"{i}. {item}")

# Tuple for fixed values

WEEKDAYS = ("Mon", "Tue", "Wed", "Thu", "Fri")
TIME_SLOTS = ("09:00", "13:00", "17:00")
WEEKDAYS, TIME_SLOTS, type(WEEKDAYS), type(TIME_SLOTS)

# Combine tuples and list
schedule_map = {}
for day in WEEKDAYS:
    match = next((item for item in weekly_schedule if item.startswith(day)), None)
    schedule_map[day] = match if match else f"{day}: (free)"
 
schedule_map
```

```python
# Dictionaries & Sets
grades = {
    "Ada": 95,
    "Alan": 88,
    "Grace": 95,
    "Linus": 79,
}
grades, type(grades) #prints: ({'Ada': 95, 'Alan': 88, 'Grace': 95, 'Linus': 79}, dict)

print("Ada's score:", grades["Ada"]) # Ada's score: 95
 
try:
    print(grades["Marie"])   # raise keyerror
except KeyError as e:
    print("KeyError:", e)  # KeyError: 'Marie'
 
print("Missing (safe):", grades.get("Marie", "N/A")) #Missing (safe): N/A

# CRUD
grades["Marie"] = 91         # create
grades["Alan"]  = 90         # update
removed = grades.pop("Linus")  # delete (returns removed value)
grades, removed  #({'Ada': 95, 'Alan': 90, 'Grace': 95, 'Marie': 91}, 79)

# iterate keys, values, items
print("Keys:", list(grades.keys()))  # Keys: ['Ada', 'Alan', 'Grace', 'Marie']
print("Values:", list(grades.values()))  # Values: [95, 90, 95, 91]

print("Items:", list(grades.items()))  # Items: [('Ada', 95), ('Alan', 90), ('Grace', 95), ('Marie', 91)]
 
for name, score in grades.items():
    print(f"{name}: {score}")

unique_scores = set(grades.values())
print(unique_scores, type(unique_scores))

# Unique Score
unique_scores.add(100)
unique_scores.discard(79)         # safe remove (no error if missing)
print("Has 95?", 95 in unique_scores)
unique_scores

# mini exercise
grades["Peter", "Ryan"] = 85
unique_scores = set(grades.values())
print("Unique Scores (collapsed):", unique_scores)

print("Safe lookup:", grades.get("Tim", "Student Not Found"))


```

```python
reverse_index = {}

for name, score in grades.items():
    # If the score is not in the new dictionary yet, create an empty list for it
    if score not in reverse_index:
        reverse_index[score] = []
    
    # Append the student's name to that score's list
    reverse_index[score].append(name)

print("Reverse Index:")
print(reverse_index)
```
