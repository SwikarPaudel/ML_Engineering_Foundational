```python
# Handling Zero division error
try:
    result = 10 / 0
except ZeroDivisionError:
    print("Error: You cannot divide a number by zero!")
    result = 0
print(f"Final Result: {result}")
```

```python
# A common issue: loading raw values from an uncleaned dataset
raw_input = "not_a_number"

try:
    # Tries to convert text to integer
    number = int(raw_input)
    #  Tries to divide by the number
    answer = 100 / number

except ValueError:
    print("Error: Could not convert the input text into a valid number.")

except ZeroDivisionError:
    print(" Error: The converted number was zero. Cannot divide by zero.")
```

```python
# else: Runs only if the code in the try block succeeded with zero errors.
# finally: Runs no matter what happens (even if the code completely crashed).
try:
    print("Opening dataset file...")
    file_status = "Success"
except Exception as e:
    print(f"Failed to read file: {e}")
else:
    print("Dataset processed cleanly with no errors!")
finally:
    print("Closing dataset file connections to clear up RAM memory.\n")
```
