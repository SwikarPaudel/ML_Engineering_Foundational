```python
# Passing parametes
def welcome_user(name):
    print(f"Hello {name}, ready to code?")

welcome_user("Mystic")
```

```python
# Return data
def calculate_total_classes(weeks, classes_per_week):
    total = weeks * classes_per_week
    return total  # Sends the answer back

# Call the function and save the answer into a variable
remaining_classes = calculate_total_classes(4, 3)
print(f"You have {remaining_classes} classes this month.")  # Prints: 12
```

```python
# Default Parameters
def setup_model(learning_rate=0.001):
    print(f"Training model with learning rate: {learning_rate}")

setup_model()         # Uses default -> Prints: 0.001
setup_model(0.05)     # Overwrites default -> Prints: 0.05
```

```python
# Positional Arguments
def power(base, exp):
    return base ** exp

result = power(2,3)
print(result)
```

```python
# Keyword Argument
def power(base, exp):
    return base ** exp
result = power(exp = 3, base = 2)  #order does not matter here
print(result)
```

```python
 # Calculate the area of rectangle
def calculate_area(length, width):
    return length*width

result = calculate_area(12, 5)
print(result)
```

```python
# Create a greeting function with a default parameter for time of day
def greet (name, time_of_day = "Morning"):
    return f"Good {time_of_day}, {name}!"

print(greet("Mystic"))
print(greet("Mystic", "Afternoon"))
print(greet("Mystic", "Evening"))
```

```python
# Write a function that returns the largest of three numbers

def find_max(a, b, c):
    if a >= b and a >= c:
        return a
    elif b >= a and b >= c:
        return b
    else:
        return c

print("Largest:", find_max(15, 42, 23))  # Prints: Largest: 42
print("Largest:", find_max(10, 5, 10))   # Prints: Largest: 10
    
```
