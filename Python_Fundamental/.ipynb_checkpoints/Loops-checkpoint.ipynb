```python
# Iterating over a simple list
classes = ["Python", "Math", "Machine Learning"]

for topic in classes:
    print(f"Now studying: {topic}")
```

```python
# Print numbers from 1 to 5
for i in range(1, 10, 2): #(start, end, steps)
    print(f"Class number: {i}")
```

```python
# A simple countdown loop
classes_left = 3

while classes_left > 0:
    print(f"You have {classes_left} classes remaining.")
    classes_left -= 1  # Decrement the counter to avoid an infinite loop

print("All classes done for the day!")
```

```python
# Example of break and continue
for num in range(1, 6):
    if num == 2:
        print("Skipping 2!")
        continue  # Skips printing '2' and goes straight to 3
        
    if num == 4:
        print("Found 4! Stopping loop entirely.")
        break     # Crashing out of the loop completely
        
    print(f"Processing number: {num}")
```

```python
# Nested Loop
for i in range(1, 6):
    for j in range(1, i+1):
        print(j, end = "")
    print()
        
```

```python
for i in range(5,0,-1):
    for j in range(1, i+1):
        print(j, end = "")
    print()
```

```python
rows = 5
for i in range(1, rows + 1):
    print(" " * (rows-i), end = "")
    for j in range(1, i+1):
        print(j, end = "")
    print()
```

```python
for i in range(1,6):
    for j in range(5, i-1, -1):
        print(j, end = "")
    print()
```
