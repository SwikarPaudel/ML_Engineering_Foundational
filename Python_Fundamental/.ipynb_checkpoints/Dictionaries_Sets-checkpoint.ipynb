```python
# Create a dictionary using curly brackets and colons
student_profile = {
    "name": "Mystic",
    "topic": "ML Engineering",
    "completed_classes": 5
}

#  Look up a value using its key
print(student_profile["name"])  # Prints: Swikar

#  Add or update a value
student_profile["completed_classes"] = 6  # Updates existing key
student_profile["status"] = "Active"       # Adds a brand new key

print(student_profile)
```

```python
# Create a set using curly brackets (no colons)
registered_ids = {101, 102, 103, 101}

# Notice that the duplicate '101' is automatically removed!
print(registered_ids)  # Prints: {101, 102, 103}

# Add an item
registered_ids.add(104)

# Fast membership testing (checking if something exists)
print(102 in registered_ids)  # Prints: True
```

```python
# Create a dictionary for a book with keys: title, author, year. Add a new key: "genre"
book = {
    "title": "Avengers",
    "author": "Avengers Writer",
    "year" : 2024
}
print(book)

book["genre"] = "Scifi"
print(book)
```

```python
# Create a set of 3 colors. Add a new color to your set . Create a second set of colors and find the intersection
color1 = { "Red", "Green", "Blue"}
print(color1)

color1.add("Yellow")

color2 = {"Green", "Blue", "Black", "White"}

# common_colors = color1.intersection(color2)
common_colors = color1 & color2
#Both works as intersections
print(common_colors)
```
